Significant Margin Drop

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

monthly_driver = (
    df.groupby(df['Order_Date'].dt.to_period('M'))
    .agg(
        Revenue=('Sales_BDT', 'sum'),
        Cost=('Cost_BDT', 'sum'),
        Profit=('Profit_BDT', 'sum'),
        Avg_Discount=('Discount_Pct', 'mean')
    )
    .reset_index()
)

monthly_driver['Order_Date'] = monthly_driver['Order_Date'].astype(str)

monthly_driver = monthly_driver.sort_values('Order_Date').reset_index(drop=True)

monthly_driver['Margin_Pct'] = (
    monthly_driver['Profit'] / monthly_driver['Revenue'] * 100
)
monthly_driver['Margin_Change_Pp'] = monthly_driver['Margin_Pct'].diff()

monthly_driver['Revenue_Growth_Pct'] = monthly_driver['Revenue'].pct_change() * 100
monthly_driver['Cost_Growth_Pct'] = monthly_driver['Cost'].pct_change() * 100

print(
    monthly_driver
    .sort_values('Margin_Change_Pp', ascending=True)
    .head(10)
    .round(2)
)
```

```python
 
problem_month = monthly_driver.sort_values('Margin_Change_Pp', ascending=True)['Order_Date'].iloc[0]

monthly_driver['Order_Date'] = pd.PeriodIndex(monthly_driver['Order_Date'], freq='M')

problem_period = pd.Period(problem_month, freq='M')
previous_period = problem_period - 1

comparison = monthly_driver[
    monthly_driver['Order_Date'].isin(
        [previous_period, problem_period]
    )
].copy()

print(comparison.round(2))
```
