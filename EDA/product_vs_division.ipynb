```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Talib\Desktop\DA - Projects\New Project Resume\Dataset\files v2\Cleaned Data\Fact_Sales.csv")
df
```

```python
customer = pd.read_csv(r"C:\Users\Talib\Desktop\DA - Projects\New Project Resume\Dataset\files v2\Cleaned Data\Dim_Customer.csv")
customer
```

```python
df_customer = df.merge(
    customer[['Customer_ID', 'Segment', 'Division', 'District']],
    on='Customer_ID',
    how='left'
)

print(df_customer.columns)
```

Regional Performance & Profitability Breakdown by Division

```python

division_analysis = (
    df_customer.groupby('Division')
    .agg(
        Revenue=('Sales_BDT', 'sum'),
        Cost=('Cost_BDT', 'sum'),
        Profit=('Profit_BDT', 'sum'),
        Quantity=('Quantity', 'sum')
    )
    .reset_index()
)

division_analysis['Margin_Pct'] = (
    division_analysis['Profit'] /
    division_analysis['Revenue'] * 100
)

division_analysis['Revenue_Share_Pct'] = (
    division_analysis['Revenue'] /
    division_analysis['Revenue'].sum() * 100
)

division_analysis['Profit_Share_Pct'] = (
    division_analysis['Profit'] /
    division_analysis['Profit'].sum() * 100
)

print(
    division_analysis
    .sort_values('Profit', ascending=False)
    .round(2)
)
```

Best Region/Division

```python
print("TOP DIVISIONS BY PROFIT")

print(
    division_analysis
    .sort_values('Profit', ascending=False)
    .head(3)
    .round(2)
)
```

Worst Margin Region

```python
print("\nLOWEST MARGIN DIVISIONS")

print(
    division_analysis
    .sort_values('Margin_Pct')
    .head(3)
    .round(2)
)
```

HIGH-REVENUE BUT LOW-MARGIN DIVISIONS

```python
high_revenue_low_margin_region = division_analysis[
    (division_analysis['Revenue'] >= division_analysis['Revenue'].median()) &
    (division_analysis['Margin_Pct'] <= division_analysis['Margin_Pct'].median())
]

print("\nHIGH-REVENUE BUT LOW-MARGIN DIVISIONS")

print(
    high_revenue_low_margin_region
    .sort_values('Revenue', ascending=False)
    .round(2)
)
```
